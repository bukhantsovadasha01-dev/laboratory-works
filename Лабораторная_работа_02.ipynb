## Задание

Необходимо вычислить значение выражения: $$Z=1 -
\frac{X}{2}+\frac{Y^2}{6}-\frac{X^3}{24}+\frac{Y^4}{120}-...$$

## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые
потребуются для выполнения задания. Независимые переменные: $X$ и
$Y$. Переменная, которая должна быть вычислена на основе
независимых: $Z$.

Следует обратить внимание, что в выражении неизвестное число слагаемых, поэтому необходимо ввести переменную $N$ для
представления количества слагаемых.

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $Z$:

```python

X = 2
Y = 3
N = 10
Z = 0

for i in range(N):
  factorial = 1
  for j in range(1, i + 1):
    factorial = factorial * j
  if i == 0:
    slag = 1
  elif i % 2 == 1:
    slag = (X ** i) / factorial
  else:
    slag = (Y ** i) / factorial
  if i % 2 == 0:
    Z += slag
  else:
    Z-= slag
print(f"Z = {Z}")
```
