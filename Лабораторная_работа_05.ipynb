## Задание

В файле pr5_1.txt содержится последовательность целых чисел. Элементы последовательности могут принимать целые значения от 0 до 10 000 включительно. Определите количество пар, в которых оба элемента меньше, чем среднее арифметическое всех чисел в файле, и хотя бы один из двух элементов оканчивается на 9. В ответе запишите два числа: сначала количество найденных пар а затем – максимальную сумму элементов таких пар. В данной задаче под парой подразумевается два идущих подряд элемента последовательности.



## Решение

```python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract5/pr5_1.txt
```

```python
# Читаем файл, разбиваем на слова и сразу превращаем в числа
data = list(map(int, open('pr5_1.txt').read().split()))

print(len(data))
print(data[:10])
```

Теперь ответим на поставленные вопросы использую список `data`

```python
# Вычислим среднее арифметическое всех чисел в файле
sr = sum(data) / len(data)
print(sr)
```

```python
# Подсчет количества подходящих пар
counter = 0
max_sum = -10000
for i in range(len(data)-1):
    # пара элементов
    x, y = data[i], data[i+1]
    # Условие 1: ОБА элемента меньше среднего
    # Условие 2: ХОТЯ БЫ ОДИН элемент оканчивается на 9
    if x < sr and y < sr and (x % 10 == 9 or y % 10 == 9):
        counter += 1
        max_sum = max(max_sum, x+y)

print(f'Количество подходящих пар = {counter}')
print(f'Максимальная сумма = {max_sum}')
```

3. Ответ: [10, 10, 10, 10, 10]

4. Нужно использовать метод insert(). Правильная команда: names.insert(0, 'Ванди').

    Почему: Метод append() добавляет элемент только в конец списка. Присваивание по индексу (например, names[0] = ...) сработает только если элемент с таким индексом уже существует. А insert(0, ...) специально создан для вставки на конкретную позицию со сдвигом остальных элементов.

5. index() - возвращает индекс (порядковый номер) первого вхождения указанного элемента в список. Если элемента нет, выдает ошибку.

    insert() - вставляет элемент на указанную позицию, сдвигая все последующие элементы вправо.

    sort() - сортирует элементы списка по возрастанию (или по убыванию, если указан параметр reverse=True).

    reverse() - меняет порядок элементов списка на обратный.
