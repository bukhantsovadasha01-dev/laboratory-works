Далее я реализовала чтение всего файла построчно.
Я использовала метод `open()`, а затем обошла файл как коллекцию с использованием цикла `for line in File`.
CSV-файл в данной работе используется как обычный текстовый файл, и в цикле `for` производится последовательный обход всех строк этого файла,
то есть текстовый файл — это коллекция его строк.
В файле много строк, поэтому я вывела только первые `n_to_print` строк, но обработала все (чтобы не возникло скрытых ошибок).

````python
F = open('p7_data_06.csv')
n_to_print = 10
for line in F:
    n_to_print -= 1
    if n_to_print > 0:
        print(line)
````


Строки перебираются, но теперь мне необходимо разделить единую строку на отдельные элементы данных.
Для этого я использовала метод `split(sep)`, который разрезает строку, используя символ `sep` в качестве разделителя.
Также я обратила внимание на тот факт, что в первой строке содержатся не данные, а заголовки столбцов.

````python
# открываем файл
F = open('p7_data_06.csv')

# читаем первую строку (заголовок) и разбиваем по запятым
header = F.readline().strip().split(',')
print(header)
print('-' * 40)

n_to_print = 10
for line in F:
    data = line.strip().split(',')

    # уменьшаем счётчик
    n_to_print -= 1

    # выводим только первые n_to_print строк
    if n_to_print > 0:
        print(data)
````

Метод `split()` вернул список отдельных строк — подстрок исходной строки.
Теперь мне необходимо сформировать структуру данных.
Для этого я сформирую список словарей, то есть все данные в процессе чтения буду построчно перемещать в список `all_targets`, при этом каждая цель будет представлена в виде словаря.

Например, первая строка в файле будет выглядеть следующим образом:

```python
example = {'Target_ID': '18-3279071', 'Param1': 784, 'Param2': 97.3,
           'Param3': 78.23, 'Category': 'Трек', 'HasWeapon': True}

````python

F = open('p7_data_06.csv')

header = F.readline().strip().split(',')

# создаём пустой список для всех целей
all_targets = []

for line in F:
    data = line.strip().split(',')

    # создаём пустой словарь для одной цели
    target = {}

    # заполняем словарь
    for i, key in enumerate(header):
        target[key] = data[i]

    # превращаем строки в числа
    target['Param1'] = int(target['Param1'])
    target['Param2'] = float(target['Param2'])
    target['Param3'] = float(target['Param3'])

    target['HasWeapon'] = target['HasWeapon'].strip().lower() == 'true'

    # добавляем цель в список
    all_targets.append(target)

print(all_targets[:10])

F.close()
````

## Решение задачи 1

Первая задача: определить среднее значение Param1 по каждой из категорий.

Я переберу все элементы в списке `all_targets` и сгруппирую их по категориям.
Для каждой категории я буду накапливать сумму Param1 и количество целей. Затем я вычислю среднее как отношение суммы к количеству.



````python

stats = {}

for target in all_targets:
    category = target['Category']

    # если категория встретилась впервые — создаём для неё запись
    if category not in stats:
        stats[category] = {'sum': 0, 'count': 0}

    # добавляем Param1 к сумме
    stats[category]['sum'] += target['Param1']
    stats[category]['count'] += 1

print("=== Среднее значение Param1 по категориям ===")
for cat, values in stats.items():
    avg = values['sum'] / values['count']
    print(f"{cat}: среднее Param1 = {avg:.2f}")
````

## Решение задачи 2

Определить сумму параметров Param2 для каждой категории.

Я создала отдельные переменные для суммы Param2 каждой категории:
`sum_param2_track` для Track, `sum_param2_plane` для Plane и `sum_param2_ship` для Ship.

Затем я прошла по всем целям в списке `all_targets` и для каждой цели проверила её категорию.
В зависимости от категории я прибавляла значение Param2 к нужной сумме.

В конце я вывела результат для каждой категории.

````python

sum_param2_track = 0
sum_param2_plane = 0
sum_param2_ship = 0


for target in all_targets:
    if target['Category'] == 'Track':
        sum_param2_track += target['Param2']
    elif target['Category'] == 'Plane':
        sum_param2_plane += target['Param2']
    elif target['Category'] == 'Ship':
        sum_param2_ship += target['Param2']


print("=== Сумма Param2 по категориям ===")
print('Track:', sum_param2_track)
print('Plane:', sum_param2_plane)
print('Ship:', sum_param2_ship)
````

## Решение задачи 3

Определить, сколько целей снабжены оружием, а сколько – нет.

Я создала две переменные-счётчика: `armed_count` для целей с оружием и `unarmed_count` для целей без оружия.

Затем я прошла по всем целям в списке `all_targets`. Для каждой цели я проверила значение поля `HasWeapon`. Если оно равно `True`, я увеличила счётчик целей с оружием на 1. Если `False` — увеличила счётчик целей без оружия на 1.

В конце я вывела оба значения и общую сумму для проверки.

````python

armed_count = 0
unarmed_count = 0


for target in all_targets:
    if target['HasWeapon'] == True:
        armed_count += 1
    else:
        unarmed_count += 1

print("=== Наличие оружия ===")
print(f"Целей с оружием: {armed_count}")
print(f"Целей без оружия: {unarmed_count}")
print(f"Всего целей: {armed_count + unarmed_count}")
````

## Решение задачи 4

Найти объект с максимальным значением Param1.

Я создала две переменные: `max_param1` для максимального значения (начинаем с нуля) и `max_id` для ID цели.

Затем я прошла по всем целям и сравнивала их Param1 с текущим максимумом. Если находила большее — запоминала новое значение и ID цели.

В конце вывела результат.

````python
max_param1 = 0
max_id = ""

for target in all_targets:
    if target['Param1'] > max_param1:
        max_param1 = target['Param1']
        max_id = target['Target_ID']

print("Максимальное Param1:", max_param1)
print("ID цели:", max_id)
````
