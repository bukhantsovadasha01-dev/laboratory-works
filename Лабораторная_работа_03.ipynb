## Задание

Напишите функцию, которая по трем переданным натуральным числам
возвращает могут ли эти числа являться членами арифметической
прогрессии. Если «да», то функция возвращает True, в противном случае– False.

## Решение

### Объявление функции

```python
def arifm_progr(a, b, c):
  nums = sorted([a, b,c])
  return (nums[1]- nums[0]) == (nums[2]- nums[1])
```

### Демонстрация использования функции

```python
x, y, z = 1, 3, 5
print(f'Числа {x}, {y}, {z} образуют арифметическую прогрессию: {arifm_progr(x, y, z)}')
x, y, z = 2, 7, 15
print(f'Числа {x}, {y}, {z} образуют арифметическую прогрессию: {arifm_progr(x, y, z)}')
x, y, z = 10, 20, 30
print(f'Числа {x}, {y}, {z} образуют арифметическую прогрессию: {arifm_progr(x, y, z)}')
x, y, z = 5, 5, 5
print(f'Числа {x}, {y}, {z} образуют арифметическую прогрессию: {arifm_progr(x, y, z)}')
```
