Реализуем чтение строки из файла

```python
mystr = open('pr6.txt').read().strip()

print(len(mystr))
print(mystr[:50])
```

Теперь ответим на поставленные вопросы, используя строку `mystr`

```python
# ищем кол-во букв D,перед которыми стоят две цифры
k = 0
for i in range(2, len(mystr)):
  if mystr[i] == 'D' and mystr[i-2] in '0123456789' and mystr[i-1] in '0123456789':
    k += 1
print(f'Количество букв D, перед которыми две цифры: {k}')
```
