## Задание

Вася составляет 6-буквенные слова из букв слова ДЕРЕВО так, что буквы на
позициях одинаковой чётности расположены в алфавитном порядке. Сколько
различных слов может составить Вася?

## Решение

Задача комбинаторная, но решить её можно и не зная теоретического материала комбинаторики. В Python существует библиотека itertools, которая может помочь перебрать все варианты перестановок букв.

**Подход к решению:**

1. Использовала функцию permutations() из библиотеки itertools для генерации всех возможных перестановок букв слова ДЕРЕВО.

2. Так как в слове две одинаковые буквы «Е», использовала set() для устранения дубликатов - оставляем только уникальные слова.

3. Для каждой перестановки проверила условие: буквы на нечётных позициях (1, 3, 5) и чётных позициях (2, 4, 6) должны идти в алфавитном порядке.

4. Считаем количество слов, удовлетворяющих условию.

```python
import itertools
alf = 'ДЕРЕВО'
number = 0
for word in sorted(set(itertools.permutations(alf))):
    word = ''.join(word)
    # буквы на нечётных позициях (1,3,5) и чётных (2,4,6) в алфавитном порядке
    if word[0] <= word[2] <= word[4] and word[1] <= word[3] <= word[5]:
        number += 1
print(number)
```
