## Задание

В файле data_v2.txt найдите минимальное число кратное 3 (под числом понимается непрерывная последовательность цифр).


## Анализ задания, получение данных

Файл data_v2.txt расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

```python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/main/data/data_v2.txt
```

Реализуем чтение всего файла в строку с использованием метоа `open()` и `read()`:

```python
f = open('data_v2.txt')
s = f.read().strip()
f.close()
print(len(s))
```

## Обработка символьной последовательности

Строки, подобно списка, являются коллекциями. Только в качестве элементов последовательности выступают символы. Для строк также используются функции нахождения длины `len()`, операция индексации `[]`

```python
# Вывод первых 10 символов строки с одновременным анализом буква это или цифра
for i in range(10):
    c = s[i]
    if c.isdigit():
        t = 'цифра'
    else:
        t = 'буква'
    print(c, t)
```

## Решение задачи

```python
# Ищем минимальное число кратное 3
target = ''
minn = None  # не знаем заранее диапазон чисел

for c in s:
    if c.isdigit():
        target += c
    else:
        if target != '':
            x = int(target)
            if x % 3 == 0:
                if minn is None or x < minn:
                    minn = x
            target = ''

# проверка посл. числа, если строка заканчивается цифрой
if target != '':
    x = int(target)
    if x % 3 == 0:
        if minn is None or x < minn:
            minn = x

print('Минимальное число кратное 3 = %d' % minn)
```

## Ответы на вопросы
1. Назначение:

    read() — считывает весь файл целиком (или указанное количество символов) и возвращает одну строку.
   
    readline() — считывает одну строку и возвращает её. При каждом вызове переходит к следующей строке.

    readlines() — читает все строки файла и возвращает их в виде списка строк.

    Сходства: все три функции читают данные из открытого файла, перемещают позицию считывания.

    Различия: возвращаемый тип данных (строка или список строк) и объём считываемой информации (весь файл / одна строка / все строки).

2. Использовать цикл for по объекту файла:

    f = open('file.txt')
   
    for line in f:
    
        print(line.strip())

    f.close()
3. Закрытие файла (f.close()) освобождает системные ресурсы, сбрасывает буферизованные данные на диск (при записи) и гарантирует, что файл корректно сохранён. Если файл не закрыть, данные могут быть потеряны, а ресурсы — заняты.

4. Позиция считывания - это указатель на текущее место в файле, откуда будет прочитан следующий символ/строка. При открытии файла в режиме чтения позиция находится в начале файла.

5. Файл открывается в режиме дозаписи - 'a' (append). Данные записываются в конец файла, существующее содержимое сохраняется.

6. Исключение - это объект, представляющий ошибку или нестандартную ситуацию, возникшую во время выполнения программы. Исключения прерывают нормальный поток выполнения и могут быть обработаны конструкцией try/except.

7. Программа завершается аварийно, выводится сообщение об ошибке (traceback) с указанием типа исключения и места, где оно произошло.

8. FileNotFoundError.

9. ValueError.

10. Файл автоматически создаётся (пустой), и запись начинается с его начала. Ошибки не возникает - режим 'a' создаёт файл, если он не существует.
