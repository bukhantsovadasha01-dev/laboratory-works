Необходимо вычислить значение выражения: $$Z=\frac{35}{f} + y \cdot
f - \frac{f + y}{4}$$ Для каждой переменной необходимо вывести ее
значение и тип.

## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые
потребуются для выполнения задания. Независимые переменные: $f, y$.
Переменная, которая должна быть вычислена на основе независимых:
$Z$

```python
f=10
y=32
print("f = ", f, ", type(f) = ", type(f))
print("y = ", y, ", type(y) = ", type(y))
```

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $Z$:

```python
Z=35/f + y*f- (f+y)/4
print('Z=',Z)
print('Type of Z: ', type(Z))
```
